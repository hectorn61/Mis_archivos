```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram
import pandas as pd
import os
from datetime import datetime

def plot_dendrogram_variables(data_numeric, timestamp, folder_path):
    """Genera un dendrograma de variables."""
    # Calcular la matriz de correlación
    corr_matrix = data_numeric.corr()

    # Calcular la matriz de vinculación
    Z = linkage(corr_matrix, method="ward")

    plt.figure(figsize=(15, 8))

    # Crear el dendrograma
    dendrogram(
        Z,
        labels=corr_matrix.columns,
        leaf_rotation=90,
    )

    # Ajustar las marcas de división principales en el eje y
    max_height = Z[-1,2] # Altura máxima del dendograma
    yticks = np.arange(0, max_height + 0.5, 0.5)
    plt.yticks(yticks)

    # Agregar líneas de división para las marcas principales
    plt.grid(axis='y', which='major', linestyle='--', alpha=0.7)

    # Título y etiquetas
    plt.title("Dendrograma de Variables")
    plt.xlabel("Variables", fontsize=16)
    plt.ylabel("Distancia (Correlación)", fontsize=16)
    plt.tight_layout()

    # Guardar la gráfica
    filename = os.path.join(folder_path, f"dendrogram_variables_{timestamp}.png")
    plt.savefig(filename, dpi=300)
    plt.close()
    print(f"Dendrograma de variables guardado en '{filename}'.")
    return filename

# Ejemplo de uso
if __name__ == "__main__":
    # Carga de datos ejemplo
    data_numeric = pd.DataFrame(np.random.rand(10, 10), columns=[f'Var_{i}' for i in range(10)])

    # Timestamp para el nombre del archivo
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

    # Carpeta donde se guardará el archivo
    folder_path = './'

    plot_dendrogram_variables(data_numeric, timestamp, folder_path)
```

```python
# -*- coding: utf-8 -*-
"""
Script para generar dendrogramas (Pearson, Spearman, Kendall, K-means)
a partir de análisis geoquímico de muestras de un depósito tipo SEDEX.
Guarda los gráficos en la carpeta 'results'.
"""

# 1. Importar librerías
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.cluster import hierarchy
import scipy.spatial.distance as ssd
from sklearn.cluster import KMeans
import os
import shutil

# 2. Cargar datos
if not os.path.isfile('Data.xlsx'):
    raise FileNotFoundError("El archivo 'Data.xlsx' no existe. Súbelo primero.")

df = pd.read_excel('Data.xlsx')
numeric_df = df.select_dtypes(include=['float64', 'int64'])

# 3. Función para plotear dendrogramas de correlación
def plot_dendrogram(corr_matrix, method_name):
    # Calcular matriz de distancias
    distance = 1 - np.abs(corr_matrix)
    distance = (distance + distance.T) / 2  # simetría
    np.fill_diagonal(distance.values, 0)    # diagonal cero
    linkage = hierarchy.linkage(ssd.squareform(distance.values, checks=False), method='average')
    plt.figure(figsize=(12, 6))
    hierarchy.dendrogram(linkage, labels=corr_matrix.columns, orientation='top')
    plt.title(f"Dendrograma ({method_name})")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig(f'dendrogram_{method_name}.png', dpi=300)
    plt.close()

# 4. Función para el dendrograma de K-means (con etiquetas)
def kmeans_dendro(data, n_clusters=7):
    # Transponer para agrupar variables
    data_t = data.T

    # Aplicar K-means
    kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
    labels = kmeans.fit_predict(data_t)
    centroids = kmeans.cluster_centers_

    # Generar un dendrograma a partir de los centroides
    linkage = hierarchy.linkage(centroids, method='ward')

    # Para cada cluster, obtener sus variables
    clusters = {}
    for i in range(n_clusters):
        clusters[i] = list(data.columns[labels == i])

    # Crear etiquetas descriptivas para los centroides
    centroid_labels = [f"Cluster {i+1}: {', '.join(clusters[i])}" for i in range(n_clusters)]

    # Plotear el dendrograma
    plt.figure(figsize=(14, 8))
    hierarchy.dendrogram(linkage, labels=centroid_labels, leaf_rotation=90, leaf_font_size=10)
    plt.title('Dendrograma de centroides K-means')
    plt.tight_layout()
    plt.savefig('kmeans_dendro.png', dpi=300)
    plt.close()

# 5. Dendrogramas de correlación
corr_methods = ['pearson', 'spearman', 'kendall']
for method in corr_methods:
    corr_matrix = numeric_df.corr(method=method)
    plot_dendrogram(corr_matrix, method.capitalize())

# 6. Ejecutar la función correcta y generar el dendrograma K-means
kmeans_dendro(numeric_df, n_clusters=7)

# 7. Organizar resultados
os.makedirs('results', exist_ok=True)
for file in [f for f in os.listdir() if f.endswith('.png')]:
    shutil.move(file, f'results/{file}')

print("✅ Dendrogramas generados y guardados en la carpeta 'results'.")
```

```python
# Importar librerías necesarias
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, accuracy_score, mean_squared_error, r2_score

# Cargar datos desde Excel (asegúrate de que Data.xlsx esté en tu entorno Colab)
df = pd.read_excel('Data.xlsx')

# Mostrar las primeras filas para inspección
print(df.head())

# Definir columnas categóricas y numéricas
cat_cols = ['Litologia', 'Color']
num_cols = df.select_dtypes(include=['number']).columns.tolist()

# Suponemos que la columna objetivo es la última columna numérica (ajusta según tu caso)
target_col = num_cols[-1]
num_cols.remove(target_col)

# Separar características y variable objetivo
X = df[cat_cols + num_cols]
y = df[target_col]

# Detectar si el problema es clasificación o regresión
# Si y es categórico (texto o pocos valores únicos), usar clasificación, sino regresión
if y.dtype == 'object' or y.nunique() < 20:
    problem_type = 'classification'
else:
    problem_type = 'regression'

print(f"Tipo de problema detectado: {problem_type}")

# Preprocesamiento: codificar variables categóricas con OneHotEncoder
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols)
    ],
    remainder='passthrough'  # deja pasar las columnas numéricas sin cambios
)

# Crear pipeline con preprocesamiento y modelo RandomForest
if problem_type == 'classification':
    model = RandomForestClassifier(n_estimators=100, random_state=42)
else:
    model = RandomForestRegressor(n_estimators=100, random_state=42)

pipeline = Pipeline(steps=[('preprocessor', preprocessor),
                           ('model', model)])

# Dividir datos en entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Entrenar modelo
pipeline.fit(X_train, y_train)

# Predecir en test
y_pred = pipeline.predict(X_test)

# Evaluar resultados
if problem_type == 'classification':
    print("Accuracy:", accuracy_score(y_test, y_pred))
    print("Reporte de clasificación:\n", classification_report(y_test, y_pred))
else:
    # Calcular MSE y luego RMSE manualmente
    mse = mean_squared_error(y_test, y_pred)
    rmse = mse ** 0.5  # Raíz cuadrada del MSE para obtener RMSE
    print("MSE:", mse)
    print("RMSE:", rmse)
    print("R2 score:", r2_score(y_test, y_pred))


# Importancia de variables
# Obtener nombres de las columnas después de OneHotEncoding
ohe = pipeline.named_steps['preprocessor'].named_transformers_['cat']
ohe_features = ohe.get_feature_names_out(cat_cols)
all_features = list(ohe_features) + num_cols

importances = pipeline.named_steps['model'].feature_importances_

# Mostrar importancia de variables ordenadas
feat_imp = pd.Series(importances, index=all_features).sort_values(ascending=False)
print("\nImportancia de variables:")
print(feat_imp)
```

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.cluster import AgglomerativeClustering, DBSCAN
from sklearn.decomposition import PCA
import umap.umap_ as umap
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Cargar datos
df = pd.read_excel('Data.xlsx')

# 2. Definir columnas
cat_cols = ['Litologia', 'Color']
num_cols = df.select_dtypes(include=['float64', 'int64']).columns.tolist()

# Suponemos que la variable objetivo es la última numérica (ajustar según caso)
target_col = num_cols[-1]
num_cols.remove(target_col)

X = df[cat_cols + num_cols]
y = df[target_col]

# 3. Preprocesamiento: OneHot para categóricas, StandardScaler para numéricas
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols),
        ('num', StandardScaler(), num_cols)
    ])

# 4. Pipeline con RandomForestRegressor
rf = RandomForestRegressor(n_estimators=200, random_state=42)
pipeline = Pipeline(steps=[('preprocessor', preprocessor), ('rf', rf)])

# 5. Entrenar modelo
pipeline.fit(X, y)

# 6. Extraer modelo y datos transformados
rf_model = pipeline.named_steps['rf']
X_transformed = pipeline.named_steps['preprocessor'].transform(X)

# 7. Calcular matriz de proximidad basada en hojas compartidas
def compute_proximity_matrix(rf, X):
    leaf_indices = rf.apply(X)
    n_samples = leaf_indices.shape[0]
    proximity = np.zeros((n_samples, n_samples))
    for tree_idx in range(leaf_indices.shape[1]):
        leaves = leaf_indices[:, tree_idx]
        leaf_to_samples = {}
        for i, leaf in enumerate(leaves):
            leaf_to_samples.setdefault(leaf, []).append(i)
        for samples_in_leaf in leaf_to_samples.values():
            for i in samples_in_leaf:
                for j in samples_in_leaf:
                    proximity[i, j] += 1
    proximity /= rf.n_estimators
    return proximity

proximity = compute_proximity_matrix(rf_model, X_transformed)
distance = 1 - proximity

# 8. Clustering jerárquico (puedes cambiar n_clusters)
n_clusters = 4
clusterer = AgglomerativeClustering(n_clusters=n_clusters, metric='precomputed', linkage='average')

labels = clusterer.fit_predict(distance)

df['Dominio_RF'] = labels

# 9. Reducción dimensional para visualización (UMAP recomendado para datos complejos)
reducer = umap.UMAP(random_state=42)
embedding = reducer.fit_transform(X_transformed.toarray() if hasattr(X_transformed, "toarray") else X_transformed)

# 10. Visualización clusters
plt.figure(figsize=(10,7))
sns.scatterplot(x=embedding[:,0], y=embedding[:,1], hue=labels, palette='tab10', s=50)
plt.title('Clusters de dominios geológicos basados en Random Forest + UMAP')
plt.xlabel('UMAP 1')
plt.ylabel('UMAP 2')
plt.legend(title='Dominio')
plt.show()

# 11. Importancia de variables
ohe = pipeline.named_steps['preprocessor'].named_transformers_['cat']
ohe_features = ohe.get_feature_names_out(cat_cols)
all_features = list(ohe_features) + num_cols
importances = rf_model.feature_importances_
feat_imp = pd.Series(importances, index=all_features).sort_values(ascending=False)
print("Importancia de variables:")
print(feat_imp)

# 12. Estadísticas descriptivas por dominio
print("\nResumen estadístico por dominio:")
print(df.groupby('Dominio_RF')[num_cols + [target_col]].mean())
```
